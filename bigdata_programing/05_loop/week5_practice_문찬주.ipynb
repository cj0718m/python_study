## 1. Boolean과 `if`


### 1-1. 비교식의 결과는 Boolean

비교 연산의 결과는 `True` 또는 `False`입니다.


````python
number = 12

print(number > 10)
print(number < 10)
print(number == 12)
print(number != 12)
````

````python
result = number > 10

print(result)
print(type(result))
````

### 1-2. `if` — 조건이 맞을 때만 실행

`if` 뒤의 조건이 `True`이면 들여쓴 코드가 실행됩니다.


````python
number = 12

if number > 10:
    print("10보다 큽니다.")

print("코드 실행 종료")
````

`number > 10`이 `True`이므로 들여쓴 `print()`가 실행됩니다.  
마지막 문장은 `if` 블록 밖에 있으므로 조건과 관계없이 실행됩니다.


### 1-3. 들여쓰기와 코드 블록

같은 수준으로 들여쓴 문장들은 하나의 `if` 블록에 속합니다.


````python
number = 8

if number > 5:
    print("조건을 만족합니다.")
    print("두 문장 모두 if 블록 안에서 실행됩니다.")

print("if 블록 밖의 문장입니다.")
````

첫 번째 조건이 `True`이면 들여쓴 두 문장이 모두 실행됩니다.  
들여쓰기가 끝나는 지점에서 `if` 블록도 끝납니다.

## 2. `if / else`


### 2-1. 두 가지 경우 중 하나 선택하기

시험 점수가 60점 이상이면 `Pass`, 그렇지 않으면 `Fail`로 판정해 봅시다.


````python
score = 73

if score >= 60:
    result = "Pass"
else:
    result = "Fail"

print(result)
````

`if / else`에서는 두 블록 중 하나만 실행됩니다.

정수 하나가 주어졌을 때 짝수인지 홀수인지 출력하세요.

힌트: 2로 나눈 나머지가 0이면 짝수입니다.


````python
number = 17

# 아래에 코드를 작성하세요.

if number%2==0:
    print("짝수입니다")
else:
    print("홀수입니다")

````

## 3. `if / elif / else`


### 3-1. 조건이 세 가지 이상의 경우

숫자가 양수인지, 0인지, 음수인지 판별해 봅시다.


````python
number = -3

if number > 0:
    print("양수")
elif number == 0:
    print("0")
else:
    print("음수")
````

`elif`는 위에서부터 차례로 검사합니다.  
처음 `True`가 된 블록 하나만 실행되고 나머지 조건은 검사하지 않습니다.


### 3-2. 여러 개의 독립적인 `if`

각 조건을 독립적으로 검사해야 할 때는 `if`를 여러 번 사용할 수 있습니다.


````python
score = 95

if score >= 60:
    print("합격")

if score >= 90:
    print("우수 성적")
````

`score = 95`는 두 조건을 모두 만족하므로 두 문장이 모두 출력됩니다.


같은 조건을 `if / elif`로 작성하면 결과가 달라집니다.


````python
score = 95

if score >= 60:
    print("합격")
elif score >= 90:
    print("우수 성적")
````

첫 번째 조건이 이미 `True`이므로 `elif` 조건은 검사하지 않습니다.


### 예제 2. 학점 판정

다음 기준으로 학점을 결정하세요.

- 90점 이상 → `A`
- 80점 이상 → `B`
- 70점 이상 → `C`
- 60점 이상 → `D`
- 60점 미만 → `F`

## 4. 조건 순서와 분기 설계


### 4-1. 조건 순서가 잘못된 경우

아래 코드는 큰 금액부터 구분하려는 코드지만 조건 순서 때문에 원하는 결과가 나오지 않습니다.


````python
amount = 700000

if amount >= 100000:
    level = "Level 1"
elif amount >= 500000:
    level = "Level 2"
elif amount >= 1000000:
    level = "Level 3"
else:
    level = "Level 0"

print(level)
````

`700000 >= 100000`이 먼저 `True`가 되므로 이후 조건은 검사하지 않습니다.


### 4-2. 더 높은 기준부터 검사하기

서로 겹치는 기준은 높은 기준부터 검사하면 각 구간을 나눌 수 있습니다.


````python
amount = 700000

if amount >= 1000000:
    level = "Level 3"
elif amount >= 500000:
    level = "Level 2"
elif amount >= 100000:
    level = "Level 1"
else:
    level = "Level 0"

print(level)
````

### 4-3. 조건의 겹침

다음 두 조건은 서로 겹칩니다. `score = 95`는 두 조건을 모두 만족합니다.


````python
score = 95

if score >= 60:
    print("합격")

if score >= 90:
    print("A등급")
````

겹치는 조건이 있다고 해서 항상 잘못된 것은 아닙니다.

- 두 결과를 모두 적용하려면 독립적인 `if`를 사용할 수 있습니다.
- 하나의 결과만 선택해야 한다면 `if / elif / else` 구조와 조건 순서를 설계해야 합니다.


### 4-4. 조건의 누락

다음 규칙에는 19세가 포함되지 않습니다.

- 13세 미만 → 어린이
- 13~18세 → 청소년
- 20세 이상 → 성인


````python
age = 19
group = "분류되지 않음"

if age < 13:
    group = "어린이"
elif age <= 18:
    group = "청소년"
elif age >= 20:
    group = "성인"

print(group)
````

`age = 19`에서는 어느 조건도 만족하지 않습니다.  
조건을 만들 때 가능한 입력 중 빠지는 구간이 없는지 확인해야 합니다.


````python
age = 19

if age < 13:
    group = "어린이"
elif age < 19:
    group = "청소년"
else:
    group = "성인"

print(group)
````

## 5. 복합 조건


### 5-1. `and` — 모든 조건을 만족

`and`는 양쪽 조건이 모두 `True`일 때만 `True`입니다.


````python
score = 85
attendance = 90

if score >= 80 and attendance >= 80:
    print("두 기준을 모두 만족합니다.")
````

### 5-2. `or` — 하나 이상 만족

`or`는 조건 중 하나 이상이 `True`이면 `True`입니다.


````python
score = 95
project_score = 70

if score >= 90 or project_score >= 90:
    print("한 기준 이상을 만족합니다.")
````

### 5-3. `not` — 조건 반전

`not`은 Boolean 값을 반대로 바꿉니다.


````python
has_penalty = False

print(has_penalty)
print(not has_penalty)

if not has_penalty:
    print("제재 내역이 없습니다.")
````

### 5-4. 여러 조건을 함께 사용하기


````python
score = 85
attendance = 90
has_penalty = False

if score >= 80 and attendance >= 80 and not has_penalty:
    print("모든 조건을 만족합니다.")
else:
    print("조건을 만족하지 않습니다.")
    
````

### 5-6. 범위 조건

Python에서는 두 비교식을 연결해 하나의 범위를 표현할 수 있습니다.


````python
score = 75

if 70 <= score < 80:
    print("70점 이상 80점 미만")
````

````python
score = 75

print(70 <= score < 80)
print(score >= 70 and score < 80)
````

### 5-7. `in` — 리스트 안의 값 확인


````python
method = "카드"
allowed = ["카드", "계좌이체", "간편결제"]

if method in allowed:
    print("사용 가능한 결제 방식")
else:
    print("사용할 수 없는 결제 방식")
````

### 5-8. `in` — 문자열 안의 문자열 확인


````python
email = "student@khu.ac.kr"

if "@khu.ac.kr" in email:
    print("학교 이메일")
else:
    print("학교 이메일이 아님")
````

### 예제 4. 윤년 판별

율리우스력의 윤년 추가 규칙은 다음과 같다. 이에 따라 1년은 365.25일이 된다.
4로 나누어 떨어지는 해는 윤년, 그 밖의 해는 평년으로 한다.

그레고리력은 다음과 같은 예외 규칙을 추가하였다. 이에 따라 1년은 365.2425일이 된다.
100으로 나누어 떨어지되 400으로 나누어 떨어지지 않는 해는 평년으로 한다.(즉 100의 배수 중 400의 배수면 윤년이고 아니면 평년. ex) 1600(윤), 1700(평), 1800(평), 1900(평), 2000(윤), 2100(평), ...


````python
year = 2024
if year % 4 == 0 and (year % 100 !=0 or year % 400 == 0):
    print(f"{year}년은 윤년입니디.")
else:
    print(f"{year}년은 평년입니다.")
````

### 예제 5. 세 값 중 최댓값 찾기

리스트에 임의의 숫자 세 개가 들어 있습니다.

```python
numbers = [37, 92, 58]
```

최댓값을 찾아 출력하는 코드를 작성하세요.

````python
numbers = [37, 92, 58]

maxnum=numbers[0]
if numbers[1]>maxnum:
    maxnum=numbers[1]
if numbers[2]>maxnum:
    maxnum=numbers[2]

print(maxnum)
   
````

## 정리

- `if`: 조건이 `True`일 때 실행
- `if / else`: 두 경우 중 하나 선택
- `if / elif / else`: 여러 경우 중 하나 선택
- 독립적인 여러 `if`: 조건을 각각 검사
- 조건 순서: 앞의 조건이 뒤 조건보다 먼저 적용됨
- 조건의 겹침과 누락을 확인
- `and`, `or`, `not`: 여러 조건 결합
- 범위 조건과 `in`
- 중첩 조건문
