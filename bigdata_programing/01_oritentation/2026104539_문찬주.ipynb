## 0. Google Colab에서 코드 열기

Google Colab은 별도 설치 없이 브라우저에서 파이썬 코드를 실행할 수 있는 환경입니다.

이 파일처럼 `.ipynb` 확장자를 가진 파일을 **노트북(notebook)** 이라고 합니다.  
노트북 안에서 코드가 들어 있는 칸 하나를 **셀(cell)** 이라고 합니다.

오늘은 다음 정도만 익히면 됩니다.

| 동작 | 방법 |
|---|---|
| 코드 셀 선택 | 실행할 셀을 클릭 |
| 현재 셀 실행 | 셀 왼쪽의 ▶ 버튼 |
| 실행 후 다음 셀로 이동 | `Shift + Enter` |
| 현재 셀만 실행 | `Ctrl + Enter` |

코드는 일반적으로 위에서 아래로 실행합니다.

### Colab 화면에서 확인할 것

1. 노트북 파일이 정상적으로 열렸는지 확인합니다.
2. 코드 셀 왼쪽에 있는 ▶ 버튼을 찾아봅니다.
3. 아래의 간단한 코드를 한 번 실행해 봅니다.

코드의 의미를 모두 이해할 필요는 없습니다.  
이번 주에는 **코드가 셀에 들어 있고, 셀을 실행하면 아래에 결과가 나타난다**는 점을 확인하는 것이 핵심입니다.

````python
print("Hello, World!")
````

---
## 1. 출력 — `print()`

`print()`는 괄호 안의 값을 화면에 보여주는 함수입니다.

아래 코드를 실행하면 각 줄의 결과가 바로 아래에 나타납니다.

````python
print("안녕하세요, 파이썬!")
print(42)
print("값:", 100)
````

문자도 출력할 수 있고, 숫자도 출력할 수 있습니다.  
쉼표(`,`)를 사용하면 여러 값을 함께 출력할 수도 있습니다.

---
## 2. 간단한 산술 연산

파이썬에서는 숫자를 이용해 계산할 수 있습니다.  
이번 주에는 아래 코드가 실행되는 모습만 확인합니다.

````python
print(10 + 3)   # 더하기
print(10 - 3)   # 빼기
print(10 * 3)   # 곱하기
print(10 / 3)   # 나누기
print(2 ** 10)  # 거듭제곱
````

`#` 뒤의 내용은 **주석(comment)** 입니다.  
사람이 코드를 이해하기 위해 적는 설명이며 실행 결과에는 영향을 주지 않습니다.

---
## 3. 변수의 기본 형태

변수는 값을 저장해 두고 이름으로 다시 사용할 때 씁니다.

```python
name = "홍길동"
age = 20
```

여기서 `=`는 수학에서의 '같다'보다 **오른쪽의 값을 왼쪽 이름에 저장한다**는 의미로 보면 됩니다.

````python
name = "홍길동"
age = 20

print(name)
print(age)
print("내년 나이:", age + 1)
````

지금은 변수의 세부 동작이나 메모리 구조까지 다루지 않습니다.  
**값에 이름을 붙여 두고 나중에 사용할 수 있다**는 정도만 확인합니다.

---
## 4. 오류 메시지 예시

코드를 작성하다 보면 오류 메시지가 나타날 수 있습니다.  
이번 주에는 오류를 직접 해결하지 않고, **오류 메시지가 어떤 모습인지**만 봅니다.

예를 들어 다음 코드는 문자열과 숫자를 `+`로 연결하려 하기 때문에 오류가 납니다.

```python
print("나이: " + 20)
```

실행하면 마지막 부분에 다음과 같은 메시지가 나타납니다.

```text
TypeError: can only concatenate str (not "int") to str
```

지금은 내용을 외울 필요가 없습니다.  
오류가 발생하면 **메시지의 마지막 줄에 오류 종류와 원인에 대한 설명이 나온다**는 점만 기억하면 됩니다.
