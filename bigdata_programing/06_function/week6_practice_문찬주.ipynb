## 1. for 문과 순회

같은 처리를 여러 값에 반복해서 적용할 때 for 문을 사용할 수 있습니다.

### for 문의 일반적인 구조

```python
for 반복변수 in 반복가능한_데이터:
    반복해서 실행할 코드
```

- **반복변수**: 매 반복마다 현재 원소를 저장하는 변수
- **반복가능한 데이터**: 리스트, 문자열, range()처럼 값을 순서대로 꺼낼 수 있는 데이터
- 콜론 `:` 다음 줄부터 **들여쓰기된 코드 블록**이 반복됩니다.

실행 흐름은 다음과 같습니다.

> 값 하나를 꺼냄 → 반복변수에 저장 → 들여쓴 코드 실행 → 다음 값


### 1-1. 반복문이 없다면

리스트의 모든 값을 출력한다고 가정해 봅시다.


````python
numbers = [4, 7, 2, 9]

print(numbers[0])
print(numbers[1])
print(numbers[2])
print(numbers[3])
````

같은 작업을 반복하고 있습니다. for 문을 사용하면 데이터의 길이가 달라져도 같은 구조로 처리할 수 있습니다.


````python
numbers = [4, 7, 2, 9]

# numbers에서 값을 하나씩 꺼내 number에 저장합니다.
for number in numbers:
    # 현재 number의 값을 출력합니다.
    print(number)
````

### 1-2. for 문 실행 순서

다음 코드는 세 번 반복됩니다.


````python
numbers = [4, 7, 2]

for number in numbers:
    print(number)
````

실행 순서는 다음과 같습니다.

```text
1회차: number = 4 → print(4)
2회차: number = 7 → print(7)
3회차: number = 2 → print(2)
더 이상 원소가 없으면 반복 종료
```

반복이 끝난 뒤 다음 코드로 진행합니다.


### 1-3. 리스트 순회


````python
names = ["Kim", "Lee", "Park"]

for name in names:
    print(name)
````

### 1-4. 문자열 순회

문자열도 한 글자씩 순서대로 꺼낼 수 있습니다.


````python
word = "Python"

for ch in word:
    print(ch)
````

## 2. range()와 enumerate()

for 문은 리스트뿐 아니라 range()가 만드는 숫자 범위도 순회할 수 있습니다.

### range()의 기본 형태

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

중요한 점은 **stop 값은 포함되지 않는다**는 것입니다.


### 2-1. range(stop)

range(5)는 0부터 4까지의 값을 만듭니다.


````python
# 0, 1, 2, 3, 4를 차례로 i에 저장합니다.
for i in range(5):
    print(i)
````

### 2-2. range(start, stop)


````python
for i in range(2, 6):
    print(i)
````

출력되는 값은 2, 3, 4, 5입니다. stop인 6은 포함되지 않습니다.


### 2-3. step 지정하기


````python
for i in range(2, 11, 2):
    print(i)
````

step이 음수이면 감소하는 범위를 만들 수 있습니다.


````python
for i in range(5, 0, -1):
    print(i)
````

### 2-4. 인덱스가 필요한 경우

리스트의 값이 아니라 **위치 번호**가 필요한 경우 range(len(...))을 사용할 수 있습니다.


````python
names = ["Kim", "Lee", "Park"]

for i in range(len(names)):
    print(i, names[i])
````

### 2-5. enumerate() — 위치와 값을 함께 사용하기

위치 번호와 실제 값이 모두 필요하다면 enumerate()를 사용할 수 있습니다.

```python
for 인덱스, 값 in enumerate(리스트):
    실행할 코드
```


````python
names = ["Kim", "Lee", "Park"]

# i에는 위치 번호, name에는 실제 원소가 저장됩니다.
for i, name in enumerate(names):
    print(i, name)
````

start=1을 지정하면 번호를 1부터 시작할 수 있습니다.


````python
names = ["Kim", "Lee", "Park"]

for i, name in enumerate(names, start=1):
    print(i, name)
````

### range(len(...))과 enumerate() 비교

두 코드 모두 위치와 값을 출력할 수 있습니다.

```python
for i in range(len(names)):
    print(i, names[i])
```

```python
for i, name in enumerate(names):
    print(i, name)
```

값과 위치를 함께 사용할 목적이라면 enumerate()가 더 직접적으로 읽힙니다.


### 예제 2. 위치를 곱해서 더하기

리스트의 각 값에 그 값의 위치를 곱한 뒤 모두 더하세요.
위치는 1부터 셉니다.

```python
numbers = [1, -2, 3, -4]
```

계산 과정:

```text
1×1 + (-2)×2 + 3×3 + (-4)×4
= 1 - 4 + 9 - 16
= -10
```

````python
numbers = [1, -2, 3, -4]

# 아래에 코드를 작성하세요.
n=0
for i,num in enumerate(numbers, start=1):
    result=i*num
    n+=result
    
print(n)
````

## 3. while 문과 반복 조건

for 문은 순회할 데이터나 반복 범위가 정해져 있을 때 자연스럽습니다.

while 문은 **조건이 True인 동안** 반복합니다.

### while 문의 일반적인 구조

```python
while 조건:
    반복해서 실행할 코드
    상태 갱신
```

### 3-1. 기본 구조


````python
# 현재 상태
count = 1

# count가 5 이하인 동안 반복합니다.
while count <= 5:
    print(count)

    # count가 변해야 언젠가 조건이 False가 됩니다.
    count += 1
````

만약 `count += 1`이 없다면 count는 계속 1이고, `count <= 5`가 계속 True이므로 반복이 끝나지 않습니다.


### 3-2. for와 while 비교

1부터 5까지 출력하는 두 코드는 같은 결과를 만듭니다.


````python
for i in range(1, 6):
    print(i)
````

````python
i = 1

while i <= 5:
    print(i)
    i += 1
````

반복 횟수나 순회할 데이터가 명확하면 for가 간결하고,
종료가 상태 변화에 의해 결정된다면 while이 적절한 경우가 많습니다.


### 3-3. while로 구현하는 유클리드 알고리즘

48과 18의 최대공약수를 구해 봅시다.

두 수 `a`, `b`가 있을 때 `a % b`의 나머지를 반복해서 사용합니다.


````python
a = 48
b = 18

# b가 0이 될 때까지 반복합니다.
while b != 0:
    remainder = a % b

    # 다음 반복에서 사용할 두 수를 갱신합니다.
    a = b
    b = remainder

print(a)
````

## 4. break와 continue

반복문은 정해진 범위나 조건이 끝날 때까지 진행됩니다.

하지만 반복 도중 실행 흐름을 바꿀 수도 있습니다.

- **break**: 반복문 전체를 즉시 종료
- **continue**: 현재 반복의 남은 코드를 건너뛰고 다음 반복으로 이동


### 4-1. break — 반복 종료


````python
numbers = [4, 7, 2, 9, 1]
target = 9

for number in numbers:
    # 현재 값이 목표값인지 확인합니다.
    if number == target:
        print("찾음")

        # 목표를 찾았으므로 반복 전체를 종료합니다.
        break
````

### 4-2. continue — 현재 값을 반복 대상에서 제외

예를 들어 다음 점수 목록에서 `-1`은 결시를 의미.

결시자는 계산에서 제외합니다.

````python
scores = [85, -1, 72, 91, -1, 88]

total = 0

for score in scores:
    # -1은 결시이므로 현재 반복에서 더 이상 처리하지 않습니다.
    if score == -1:
        continue

    # 이 아래 코드는 실제 점수가 있는 경우에만 실행됩니다.
    print("점수:", score)

    if score >= 90:
        print("A")
    elif score >= 80:
        print("B")
    else:
        print("C 이하")

    total += score

print("총점:", total)
````

같은 동작을 if로도 구현할 수 있습니다.

````python
for score in scores:
    if score != -1:
        print("점수:", score)

        if score >= 90:
            print("A")
        elif score >= 80:
            print("B")
        else:
            print("C 이하")
            
````

continue가 없어도 대부분의 코드는 if/else로 다시 작성할 수 있습니다.

다만 제외 조건이 여러 개이거나 아래의 처리 코드가 길어질수록,
continue를 사용하면 **제외할 경우를 먼저 처리하고 본 처리 코드를 덜 중첩되게 작성**할 수 있습니다.


### 4-3. break와 continue의 차이

| 문법 | 의미 | 다음 반복 |
|---|---|---|
| break | 반복 전체 종료 | 실행하지 않음 |
| continue | 현재 회차의 남은 코드만 생략 | 다음 회차 계속 |


### 예제 4. while 문으로 구현하기

결시자를 제외했던 위의 for문 코드를 while 문으로 구현합니다.


````python
scores = [85, -1, 72, 91, -1, 88]
total=0
i=0
while i<len(scores):
    if scores[i] != -1:
        total +=scores[i]
        print('점수:',scores[i])
        if scores[i]>=90:
            print("A")
        elif scores[i]>=80:
            print("B")
        else:
            print("C이하")
    
    i+=1
    
print('총점:',total)

    
````

## 5. 중첩 반복문

반복문 안에 또 다른 반복문을 넣을 수 있습니다.

### 일반적인 구조

```python
for 바깥변수 in 바깥범위:
    for 안쪽변수 in 안쪽범위:
        반복할 코드
```

바깥 반복이 한 번 실행될 때마다 안쪽 반복은 처음부터 끝까지 실행됩니다.


````python
for i in range(2):
    for j in range(3):
        print(i, j)
````

실행 순서는 다음과 같습니다.

```text
i = 0 → j = 0, 1, 2
i = 1 → j = 0, 1, 2
```

총 실행 횟수는 `2 × 3 = 6`회입니다.

## 6. 반복 흐름으로 알고리즘 설계하기

코드를 바로 작성하기 전에 다음 질문을 먼저 정리하면 반복문을 설계하기 쉽습니다.

1. **무엇을 반복하는가?**  
   리스트의 값, 숫자 범위, 특정 상태 등

2. **반복하면서 무엇을 판단하는가?**  
   조건 만족 여부, 목표값과 일치 여부 등

3. **반복 과정에서 무엇을 저장하거나 갱신하는가?**  
   합계, 개수, 최댓값, 현재 위치 등

4. **언제 반복을 끝내는가?**  
   끝까지 순회, 목표 발견, 조건이 False가 됨 등

5. **일부 경우를 건너뛰어야 하는가?**  
   그렇다면 continue를 고려할 수 있음


### 예제 5. 가장 긴 연속 구간

문자열 `"HHTTHTHHTHHHHHTH"`에서 `H`가 가장 길게 연속된 길이를 구하세요.

이 문제에서는 두 상태를 따로 관리합니다.

- `current`: 현재 이어지고 있는 H의 길이
- `longest`: 지금까지 발견한 가장 긴 H 구간

```text
문자를 하나씩 확인

H라면
    current를 1 증가
    current가 longest보다 크면 longest 갱신

H가 아니라면
    current를 0으로 초기화
```

````python
text = "HHTTHTHHTHHHHHTH"

current = 0
longest = 0

for i in text:
    if i=="H":
        current+=1
        if current>longest:
            longest=current
    else:
        current=0
print("가장 긴 H 구간:", longest)
````

### 예제 6. 버블 정렬

다음 리스트를 오름차순으로 정렬한다고 가정합니다.

```python
numbers = [5, 2, 4, 1]
```

버블 정렬은 **서로 인접한 두 값을 비교하고, 왼쪽 값이 더 크면 위치를 바꾸는 과정**을 반복합니다.

첫 번째 순회의 변화:

```text
[5, 2, 4, 1]

5와 2 비교 → 교환
[2, 5, 4, 1]

5와 4 비교 → 교환
[2, 4, 5, 1]

5와 1 비교 → 교환
[2, 4, 1, 5]
```

한 번의 순회가 끝나면 가장 큰 값 하나의 위치가 뒤쪽에서 확정됩니다.


````python
numbers = [7, 3, 9, 1, 5]

# 버블 정렬을 직접 구현하세요.
for j in range(len(numbers)-1):
    for i in range(len(numbers)-1):
        if numbers[i]>numbers[i+1]:
            numbers[i],numbers[i+1]=numbers[i+1],numbers[i]

print(numbers)
````

### 예제 7. 소수 판별

2 이상의 정수 `n`이 주어졌을 때 소수인지 판별하세요.

````python
n = 97

# 소수 판별 코드를 작성하세요.
for i in range(2,n):
    if n%i==0:
        j='not prime num'
        break
    else:
        j='prime num'
        
print(f"{n} is {j}.")
````
