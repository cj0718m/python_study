## 1. 함수의 기본

함수(function)는 **특정 작업을 수행하는 코드 블록에 이름을 붙인 것**입니다.

이미 사용한 `print()`, `len()`, `sum()`, `input()`등은 파이썬이 제공하는 **내장 함수**입니다.
```python
def 함수이름(매개변수):
    함수에서 실행할 코드
```

- `def`는 함수를 **정의**합니다. 정의만으로 본문이 실행되지는 않습니다.
- `함수이름(...)`처럼 **호출**할 때 본문이 실행됩니다.
- 들여쓰기로 함수에 속하는 코드를 구분합니다.


### 1-1. 함수 정의와 호출

상품 가격에 10% 할인을 적용하는 계산을 세 번 반복하는 상황을 생각해 봅시다.

````python
print(int(12000 * 0.9))
print(int(25000 * 0.9))
print(int(40000 * 0.9))
````

할인율이 바뀌면 세 줄을 모두 수정해야 합니다. 계산 규칙을 함수 한 곳에 정의해 보겠습니다.

````python
# 함수 정의
def show_discount(price):
    """상품 가격에 10% 할인을 적용한 금액을 출력한다."""
    print(int(price * 0.9))
````

````python
# 함수 호출
show_discount(12000)
show_discount(25000)
show_discount(40000)
````

### 1-2. Docstring — 함수를 정의할 때 함께 작성하는 설명

함수 본문의 **첫 문장**에 작성하는 문자열을 **Docstring**이라고 합니다. 보통 세 개의 큰따옴표 `""" ... """`를 사용합니다.

- **목적:** 함수가 무엇을 하는지
- **Args:** 각 매개변수의 의미와 자료형
- **Returns:** 어떤 값을 반환하는지 (또는 반환값이 없는지)

`Args`와 `Returns`는 널리 쓰이는 **문서화 형식의 예시**이며, 모든 함수에 반드시 써야 하는 문법은 아닙니다. 간단한 함수는 한 줄 설명으로도 충분합니다.


````python
def print_taxed_price(price):
    """상품에 10% 세금을 적용한 가격을 출력한다.

    Args:
        price (int | float): 세금 적용 전 가격

    Returns:
        None: 화면에만 출력하고 값을 반환하지 않는다.
    """
    print(price * 1.1)

print_taxed_price(10000)
````

### 1-3. `help()`와 `.__doc__`으로 설명 확인하기

파이썬은 Docstring을 함수의 정보로 보관합니다.

- `함수.__doc__`: Docstring 문자열을 직접 확인
- `help(함수)`: 함수의 도움말을 출력 (이름 뒤에 `()`를 붙이지 않음)


````python
print(print_taxed_price.__doc__)
````

````python
help(print_taxed_price)
````

**Docstring과 주석의 차이**

- Docstring: 함수를 사용하는 사람이 알아야 할 **목적·입력·출력**을 설명
- `#` 주석: 코드의 특정 구현을 설명하거나 이유를 메모

Docstring은 함수 본문의 첫 문장에 있어야 `help()`와 `.__doc__`에서 확인됩니다.

## 3. 반환값 (`return`)

`return`은 값을 **호출한 쪽에 전달**합니다. 그리고 **실행되는 순간 함수가 즉시 종료**됩니다.

```python
def 함수이름(매개변수):
    결과 = 계산식
    return 결과
```

### 3-1. 계산 결과 반환하기

````python
def add(a, b):
    result = a + b
    return result
````

````python
value = add(3, 5)
print(value)
````

````python
print(add(10, 20))
````

`add(3, 5)` 호출 → 내부 계산 → `return 8` → 호출 위치의 변수 `value`에 8을 저장합니다.

### 3-2. `return`은 함수를 즉시 종료한다

`return` 아래에 코드가 있더라도 **실제로 그 경로를 지나지 않으면 실행되지 않습니다.**

````python
def stock_status(stock):
    if stock == 0:
        return "품절"
    print("재고 확인 완료")
    return "구매 가능"
````

````python
print(stock_status(0))  # 재고 확인 메시지는 실행되지 않음
````

````python
print(stock_status(5))
````

### 3-3. 여러 `if` 분기에서 서로 다른 값 반환하기

위에서부터 조건을 확인하고, **처음 실행되는 `return`**에서 함수가 종료됩니다. 이후 조건문을 확인하지 않습니다.

````python
def evaluate(score):
    if not (0 <= score <= 100):
        return "오류"
    if score >= 90:
        return "우수"
    if score >= 60:
        return "통과"
    return "재학습"

for score in [-5, 95, 75, 50]:
    print(score, "→", evaluate(score))
````

### 3-4. `print()`와 `return`의 차이

- `print()`: 화면에 보이도록 출력
- `return`: 함수를 호출한 곳으로 결과 전달
- 명시적인 `return` 없이 끝나면 `None` 반환

````python
def double_print(x):
    print(x * 2)

result = double_print(5)   # 내부에서 10 출력
print("반환값:", result)  # 별도의 return이 없으므로 None
````

````python
def double_return(x):
    return x * 2

result = double_return(5)
print("반환값:", result)
print("다음 계산:", result + 1)
````

### 3-6. `lambda` — 짧은 함수 표현식(일회용)

한 가지 **표현식만** 계산하는 짧은 함수를 만들 수 있습니다. 여러 문장, 복잡한 분기, 상세한 설명이 필요한 함수는 `def`를 사용합니다.

````python
def add_fee(price):
    return price * 1.05

add_fee_lambda = lambda price: price * 1.05

print(add_fee(10000))
print(add_fee_lambda(10000))
````

### 예제 2. 윤년 판별 — 조건에 따라 다르게 `return`

연도를 전달받아 윤년이면 `True`, 아니면 `False`를 반환하는 `is_leap_year(year)` 함수를 작성하세요.

윤년은 다음 규칙으로 판별합니다.

1. **400의 배수**이면 윤년입니다.
2. 그렇지 않고 **100의 배수**이면 윤년이 아닙니다.
3. 그렇지 않고 **4의 배수**이면 윤년입니다.
4. 나머지는 윤년이 아닙니다.

| 입력 | 반환값 |
|---|---|
| `2024` | `True` |
| `1900` | `False` |
| `2000` | `True` |
| `2023` | `False` |


````python
def is_leap_year(year):
    """연도가 윤년인지 판별해 True 또는 False를 반환한다."""
    # 여기에 코드를 작성하세요
    if year%400==0:
        return True
    elif year%100==0:
        return False
    elif year%4==0:
        return True
    else:
        return False
````

````python
print(is_leap_year(2024))  # 기대: True
print(is_leap_year(1900))  # 기대: False
print(is_leap_year(2000))  # 기대: True
print(is_leap_year(2023))  # 기대: False
````

## 4. 변수의 범위 (Scope)

- **지역 변수(local)**: 함수 안에서 만들고 사용하는 변수
- **전역 변수(global)**: 함수 밖(파일 수준)에서 정의한 변수
- 필요한 데이터는 보통 **인자로 받고, 결과는 `return`으로 전달**합니다.

### 4-1. 매개변수와 지역 변수는 함수 내부에서 사용

````python
score = 70

def add_bonus(score):
    result = score + 10
    return result

new_score = add_bonus(score)
print("새 점수:", new_score)
print("원래 점수:", score)
````

함수 매개변수 `score`는 함수 밖의 `score`와 이름이 같더라도 **별개의 지역 이름**입니다. 함수 밖 변수의 값은 바뀌지 않습니다.

### 4-2. 이름 가리기(Name Shadowing)

함수 안에서 전역 변수와 **같은 이름의 지역 변수**를 만들면, 함수 내부에서는 지역 변수가 먼저 사용됩니다. 이 현상을 **Name Shadowing**이라고 합니다.


````python
discount_rate = 0.10          # 전역 변수: 기본 할인율 10%

def promotion_price(price):
    discount_rate = 0.20      # 같은 이름의 지역 변수: 특별 할인율 20%
    return price * (1 - discount_rate)

print("특별가:", promotion_price(10000))  # 8000.0
print("기본 할인율:", discount_rate)      # 0.1 (전역 변수는 그대로)
````

`promotion_price()` 안에서는 지역 변수 `discount_rate = 0.20`이 사용됩니다. 함수 밖에서는 여전히 전역 변수 `discount_rate = 0.10`이 사용됩니다. **같은 이름이라도 다른 스코프의 변수**입니다.

### 4-3. 지역 변수는 함수 밖에서 사용할 수 없음


````python
def calculate():
    temp = 10
    return temp * 2

print(calculate())  # 20
# print(temp)       # 오류! 함수 밖에서 temp를 사용할 수 없습니다.
````

`print(temp)`의 주석(`#`)을 지우고 실행하면 `NameError`가 발생합니다.

### 4-4. 함수 밖의 전역 변수를 읽는 경우

함수 안에 같은 이름의 지역 변수가 없다면 전역 변수를 읽을 수 있습니다. **읽기만 할 때는 `global` 선언이 필요하지 않습니다.**


````python
tax_rate = 0.1  # 전역 변수

def total_price(price):
    return price * (1 + tax_rate)

print(total_price(10000))  # 11000.0
````

### 4-5. `global` — 함수 밖 변수를 직접 수정하기

함수 안에서 전역 변수 이름에 값을 **다시 할당**하려면 `global`을 선언합니다. 전역 변수를 여러 함수가 바꾸면 **함수 호출 순서에 따라 결과가 달라져** 동작 추적이 어려워질 수 있습니다.


````python
completed_orders = 0

def record_order():
    global completed_orders
    completed_orders += 1

record_order()
record_order()
print(completed_orders)  # 2
````

## 5. 함수와 알고리즘

조건문·반복문·탐색 알고리즘을 **재사용 가능한 함수**로 만들고, 필요할 때 다른 함수 안에서 호출합니다.

핵심 흐름: **입력 정의 → 역할 분리 → 처리 → 반환 → 테스트**

- **함수화:** 반복하거나 독립적으로 사용할 계산을 이름 붙여 정의하기
- **모듈화:** 큰 문제를 서로 다른 역할의 작은 함수로 나눠 조합하기
- **좋은 함수:** 목적이 명확하고, 입력·출력이 예측 가능하며, 따로 검증할 수 있는 함수

### 5-1. 조건을 검사하는 함수(boolean을 반환)

````python
def is_valid(score):
    return 0 <= score <= 100

for s in [90, -1, 75, 101]:
    if is_valid(s):
        print(s)
````

- `is_valid()`: 점수가 유효한지 판정하는 **함수**
- `for`: 점수를 하나씩 확인하는 **반복문**
- `if`: 함수의 반환값이 `True`일 때만 출력하는 **조건문**

### 5-2. 여러 값에 함수를 적용하기

함수로 판단 규칙을 만들고, 반복문에서 그 함수를 호출할 수 있습니다.

````python
def is_even(n):
    return n % 2 == 0

numbers = [3, 4, 7, 10, 12]

for number in numbers:
    if is_even(number):
        print(number)
````

### 5-3. 함수 안에서 반복하고 결과 반환하기

이번에는 짝수의 **개수**를 반환해 봅시다.

````python
def count_even(numbers):
    count = 0
    for number in numbers:
        if number % 2 == 0:
            count += 1
    return count

print(count_even([3, 4, 7, 10, 12]))  # 3
print(count_even([1, 3, 5]))         # 0
print(count_even([]))                # 0
````

### 예제 3. 합격한 학생 수 세기

점수 목록을 전달하면 **60점 이상인 학생 수**를 반환하는 `count_pass(scores)` 함수를 작성하세요.

구현 전에 다음 네 가지를 정리합니다.

1. **입력**: `scores` 리스트
2. **처리**: `for`로 순회하며 `score >= 60`일 때 개수 증가
3. **반환**: 합격한 학생 수 (`int`)
4. **테스트**: 모두 합격 / 일부 합격 / 빈 리스트

| 입력 | 기대 반환값 |
|---|---:|
| `[60, 75, 40, 90]` | 3 |
| `[10, 20, 30]` | 0 |
| `[]` | 0 |

`print()`는 **함수 밖에서만** 사용하고, 함수는 개수를 `return`하세요.

````python
def count_pass(scores):
    count = 0
    # 아래에 반복문과 조건문을 작성하세요.
    for i in scores:
        if i >= 60:
            count+=1
    return count

print(count_pass([60, 75, 40, 90]))
print(count_pass([10, 20, 30]))
print(count_pass([]))
````

### 예제 4. 가장 긴 연속 상승 구간

매일 기록한 가격 목록에서 **가격이 전날보다 계속 상승한 가장 긴 연속 구간의 길이(일수)**를 구하세요.

```python
prices = [10, 12, 14, 11, 13, 15, 17, 16]
```

- `[10, 12, 14]`는 3일 연속 상승 구간입니다.
- `[11, 13, 15, 17]`는 4일 연속 상승 구간입니다.
- 따라서 이 예제의 답은 **4**입니다.

구현할 함수: `longest_increasing_run(prices)`

**계산 규칙**

1. 빈 리스트를 받으면 `0`을 반환합니다.
2. 데이터가 하나만 있으면 구간 길이는 `1`입니다.

```

````python
def longest_increasing_run(prices):
    # 함수를 완성하세요
    current=1
    longest=1
    for i in range(1,len(prices)):
        if prices[i]>prices[i-1]:
            current+=1
        else:
            current=1
        if current>longest:
            longest=current
    if len(prices)==0:
                longest=0
     
    return longest

print(longest_increasing_run([10, 12, 14, 11, 13, 15, 17, 16]))  # 기대: 4
print(longest_increasing_run([4, 6, 8, 3, 5]))                 # 기대: 3
print(longest_increasing_run([5, 5, 5]))                       # 기대: 1
print(longest_increasing_run([]))                              # 기대: 0
````

### 예제 5. 버블 정렬 함수

6주차에 배운 **버블 정렬**을 함수로 구현합니다. `bubble_sort(numbers)`는 **오름차순으로 정렬된 새 리스트**를 반환합니다.

**알고리즘**

1. `result = numbers.copy()`로 원본을 보존합니다.
2. **바깥 반복문**: 아직 정렬되지 않은 구간을 줄여가며 순회합니다.
3. **안쪽 반복문**: 인접한 두 값을 비교하고, 앞의 값이 크면 교환합니다.
4. 가장 큰 값이 오른쪽 끝으로 이동합니다.
5. 모든 순회를 마친 뒤 `result`를 `return`합니다.

```text
[5, 2, 4, 1]
첫 순회: [2, 5, 4, 1] → [2, 4, 5, 1] → [2, 4, 1, 5]
이후 [1, 2, 4, 5] 완성
```

여기서 `bubble_sort()`가 정렬과 반환만 담당하도록 하고, 출력은 **호출한 곳에서** 합니다.


````python
def bubble_sort(numbers):
    """입력 리스트를 바꾸지 않고 오름차순 정렬된 새 리스트를 반환한다."""
    result = numbers.copy() # copy를 하지 않으면 함수 안에서의 변경이 원본에 반영됨
    n = len(result)

    for i in range(n - 1):
        for j in range(n - 1 - i):
            if result[j] > result[j + 1]:
                result[j], result[j + 1] = result[j + 1], result[j]

    return result

values = [5, 2, 4, 1]
print(bubble_sort(values))      # [1, 2, 4, 5]
print(values)                   # [5, 2, 4, 1] (원본 유지)
print(bubble_sort([3, 1, 3]))  # [1, 3, 3]
print(bubble_sort([]))         # []
````

### 예제 6. 두 함수를 조합해 소수 개수 세기 — 풀이 예제

숫자 하나가 소수인지 판별하는 함수와, 리스트에서 소수의 개수를 세는 함수를 **분리**합니다.

- `is_prime(n)`: `n`이 소수인지 `True`/`False`를 반환
- `count_primes(numbers)`: 각 원소에 `is_prime()`을 적용해 소수의 개수를 반환

**단계별 풀이**

1. `n < 2`이면 소수가 아니므로 바로 `False`를 반환합니다.
2. 2부터 `n - 1`까지 나누어떨어지는 수가 있으면 즉시 `False`를 반환합니다.
3. 끝까지 약수가 없으면 `True`를 반환합니다.
4. `count_primes()`에서는 소수 판별을 중복 작성하지 않고 `is_prime()`을 **호출**합니다.

`[2, 3, 4, 9, 11, 1, 17]` → 소수는 `2, 3, 11, 17` → 결과 **4**


````python
def is_prime(n):
    """정수 n이 소수이면 True, 아니면 False를 반환한다."""
    if n < 2:
        return False
    for divisor in range(2, n):
        if n % divisor == 0:
            return False
    return True


def count_primes(numbers):
    """리스트에 들어 있는 소수의 개수를 반환한다."""
    count = 0
    for number in numbers:
        if is_prime(number):   # 위의 함수를 재사용
            count += 1
    return count

print(is_prime(2))                       # True
print(is_prime(9))                       # False
print(is_prime(97))                      # True
print(count_primes([2, 3, 4, 9, 11, 1, 17]))  # 4
print(count_primes([-3, 0, 1, 2]))      # 1
print(count_primes([]))                  # 0
````

**모듈화 이유?**

- `is_prime()`은 숫자 **하나**의 판별만 담당하므로 다른 문제에서도 재사용할 수 있습니다.
- `count_primes()`는 리스트 순회와 **개수 세기**만 담당합니다.
- 오류가 나면 **소수 판별**과 **개수 누적** 중 어느 부분인지 따로 검사할 수 있습니다.


### 예제 7 — 웹사이트 방문 데이터 처리

하루 방문자 수 데이터에 입력 오류 `-1`이 섞여 있습니다. 유효한 데이터만 남기고, 목표 방문자 수를 달성한 날을 세어 보겠습니다.

```python
visits = [120, -1, 180, 210, 0]
```

단순히 하나의 긴 함수에 모두 작성하는 대신 **세 가지 역할**로 나눕니다.

1. `filter_valid_visits(visits)`: 음수를 제외한 새로운 리스트 반환
2. `count_goal_days(visits, goal)`: 목표 이상인 날짜 수 반환
3. `summarize_visits(visits, goal)`: 앞의 두 함수를 연결해 결과 문자열 반환

**핵심:** 각 함수가 하나의 역할을 가지며 **이전 함수의 반환값이 다음 함수의 인자**가 됩니다.


````python
def filter_valid_visits(visits):
    """음수인 잘못된 방문자 수를 제외하고 새 리스트를 반환한다."""
    valid = []
    for count in visits:
        if count >= 0:
            valid.append(count)
    return valid


def count_goal_days(visits, goal):
    """방문자 수가 goal 이상인 날짜의 개수를 반환한다."""
    days = 0
    for count in visits:
        if count >= goal:
            days += 1
    return days


def summarize_visits(visits, goal):
    """유효 일수, 목표 달성 일수, 평균 방문자 수를 문자열로 반환한다."""
    valid = filter_valid_visits(visits)
    if not valid:
        return "유효한 데이터 없음"  # 조기 종료
    achieved = count_goal_days(valid, goal)
    average = sum(valid) / len(valid)
    return f"유효 {len(valid)}일 / 목표 달성 {achieved}일 / 평균 {average:.1f}회"

raw_visits = [120, -1, 180, 210, 0]
print(summarize_visits(raw_visits, 150))  # 유효 4일 / 목표 달성 2일 / 평균 127.5회
print(summarize_visits([-1, -5], 150))   # 유효한 데이터 없음
print("원본:", raw_visits)               # [120, -1, 180, 210, 0]
````

**정리:** `filter_valid_visits()`만 바꾸면 새로운 데이터 정제 기준을 적용할 수 있습니다. `count_goal_days()`는 다른 리스트에도 그대로 사용할 수 있습니다. 함수 분해는 코드 줄 수가 아니라 **변경·재사용·테스트의 용이성**을 위한 것입니다.

### 예제 8. 쇼핑몰 주문 금액 계산

다음 규칙을 함수로 나눠 구현하세요.

- 상품 가격 목록의 합계를 구합니다.
- 회원이면 합계에 **10% 할인**을 적용합니다.
- 할인 후 금액이 **30,000원 이상**이면 배송비 `0원`, 그보다 적으면 배송비 `3,000원`입니다.
- 최종 결제액은 **할인 후 상품 금액 + 배송비**입니다.
- 가격 리스트가 비어 있으면 주문이 없는 것으로 보고 **즉시 0**을 반환합니다.

작성할 함수:

1. `apply_member_discount(amount, is_member)`: 할인 후 금액 반환
2. `calculate_shipping(amount)`: 배송비 반환
3. `calculate_order_total(prices, is_member)`: `sum(prices)` 계산 후 위 두 함수를 **호출하여** 최종 금액 반환

**주의:** 배송비 기준은 할인 **전**이 아니라 할인 **후** 금액에 적용합니다. 빈 장바구니는 먼저 `return 0`으로 처리하세요. 각 함수의 첫 줄에 한 줄 Docstring을 적어 보세요.

| 함수 호출 | 기대 반환값 |
|---|---:|
| `calculate_order_total([12000, 8000, 9000], True)` | `29100.0` |
| `calculate_order_total([20000, 15000], False)` | `35000` |
| `calculate_order_total([], False)` | `0` |


````python
def apply_member_discount(amount, is_member):
    """TODO: 회원이면 합계에서 10%할인"""
    # 아래에 코드를 작성하세요.
    if is_member==True:
        afterdiscount=amount*0.9
        return afterdiscount
    else:
        return amount
    
    


def calculate_shipping(amount):
    """TODO: 할인 후 금액이 30000원 이상이면 무료배송,그보다 적으면 3000원"""
    # 아래에 코드를 작성하세요.
    if amount>=30000:
        delivery=0
        return delivery
    else:
        delivery=3000
        return delivery
    


def calculate_order_total(prices, is_member):
    """TODO: 가격합계->회원이면 10%할인->할인후 금액 판단 후 배송비 판단->최종 금액 계산"""
    # 0. 빈 가격 리스트면 즉시 return 0
    if len(prices)==0:
        return 0
    # 1. 상품 합계
    amount=sum(prices)
    # 2. apply_member_discount() 호출
    discount=apply_member_discount(amount,is_member)
    # 3. calculate_shipping() 호출
    delivery_fee=calculate_shipping(discount)
    # 4. 최종 금액 return
    result=discount+delivery_fee
    return result
  

print(calculate_order_total([12000, 8000, 9000], True))   # 기대: 29100.0
print(calculate_order_total([20000, 15000], False))       # 기대: 35000
print(calculate_order_total([], False))                   # 기대: 0
````
