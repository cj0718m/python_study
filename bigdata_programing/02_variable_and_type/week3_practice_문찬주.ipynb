## 1. 문자열의 구조 (String)

문자열은 텍스트를 나타내는 `str` 자료형입니다. Week 2에서는 문자열을 하나의 값으로 다뤘다면, 이번 주에는 문자열 안의 **각 문자에 위치가 있다**는 점을 이용합니다.

### 1-1. 문자열의 길이 `len()`

`len(문자열)`은 문자열에 포함된 문자의 개수를 반환합니다. 공백도 하나의 문자로 셉니다.

````python
text = "Python"
print(text)
print(type(text))
````

````python
print(len(text))
````

````python
message = "Hello Python"
print(len(message))
````

`"Hello Python"`에는 `Hello` 5글자, 공백 1개, `Python` 6글자가 있으므로 길이는 12입니다.

### 1-2. 인덱싱 (Indexing)

문자열의 각 문자에는 위치 번호인 **인덱스(index)**가 있습니다. Python의 인덱스는 0부터 시작합니다.

`"Python"`의 위치는 다음과 같습니다.

| 문자 | P | y | t | h | o | n |
|---|---:|---:|---:|---:|---:|---:|
| 인덱스 | 0 | 1 | 2 | 3 | 4 | 5 |

````python
text = "Python"
print(text[0])
````

````python
print(text[1])
print(text[3])
````

`text[0]`은 첫 번째 문자, `text[5]`는 여섯 번째 문자를 뜻합니다. 첫 위치가 0이라는 점에 주의하세요.

### 1-3. 음수 인덱스

음수 인덱스를 사용하면 문자열의 뒤에서부터 접근할 수 있습니다. 마지막 문자는 `-1`입니다.

| 문자 | P | y | t | h | o | n |
|---|---:|---:|---:|---:|---:|---:|
| 음수 인덱스 | -6 | -5 | -4 | -3 | -2 | -1 |

````python
print(text[-1])
print(text[-2])
````

마지막 문자가 필요할 때 문자열 길이를 계산해서 `text[5]`처럼 적는 대신 `text[-1]`을 사용할 수 있습니다.

### 1-4. 인덱스 범위

존재하지 않는 위치를 직접 지정하면 `IndexError`가 발생합니다. `"Python"`의 길이는 6이므로 사용할 수 있는 양수 인덱스는 0부터 5까지입니다.

````python
print(len(text))
````

아래 코드는 실행하면 오류가 발생합니다. 오류 메시지에서 `IndexError`를 확인해 보세요.

## 2. 인덱싱과 슬라이싱 (Indexing & Slicing)

인덱싱은 문자 하나를 꺼낼 때 사용합니다. 여러 문자를 연속해서 꺼낼 때는 **슬라이싱(slicing)**을 사용합니다.

### 2-1. 슬라이싱 기본

형식은 `문자열[시작:끝]`입니다.

- 시작 위치는 **포함**합니다.
- 끝 위치는 **포함하지 않습니다**.

따라서 `text[1:4]`는 인덱스 1, 2, 3의 문자를 가져옵니다.

````python
text = "Python"
print(text[1:4])
````

````python
print(text[0:2])
print(text[2:5])
````

### 2-2. 시작 또는 끝 위치 생략하기

- `[:끝]`: 처음부터 끝 위치 직전까지
- `[시작:]`: 시작 위치부터 마지막까지
- `[-n:]`: 마지막 n개 문자

````python
student_id = "2026123456"
print(student_id[:4])
````

````python
print(student_id[4:])
````

````python
print(student_id[-4:])
````

### 2-3. 문자열은 직접 수정할 수 없다

문자열은 각 위치의 문자를 직접 바꿀 수 없는 **불변(immutable)** 자료형입니다.

````python
text = "Python"
print(text)
````

아래 코드는 `TypeError`가 발생하므로 주석 처리되어 있습니다.

````python
# text[0] = "J"
````

일부를 바꾸고 싶다면 기존 문자열의 필요한 부분을 이용해 **새 문자열**을 만듭니다.

````python
new_text = "J" + text[1:]
print(new_text)
````

````python
print(text)
````

`new_text`는 `Jython`이지만 원래 변수 `text`의 값은 여전히 `Python`입니다.

### 예제 1. 휴대전화 번호 가리기

다음 휴대전화 번호에서 **앞 세 자리와 뒤 네 자리만 남기고**, 가운데 네 자리는 `****`로 가려서 출력하세요.

````python
print(new_phone)
````

## 3. 문자열 메서드 (String Methods)

메서드는 특정 객체에 연결되어 그 객체를 처리하는 기능입니다. 문자열 뒤에 점 `.`을 찍고 메서드 이름을 적습니다.

```python
문자열.메서드()
```

### 3-1. `strip()` — 앞뒤 공백 제거

사용자가 입력한 값이나 외부 데이터에는 불필요한 공백이 들어 있을 수 있습니다. `strip()`은 문자열의 **앞뒤 공백**을 제거한 새 문자열을 반환합니다.

````python
name = "   홍길동   "
print("[" + name + "]")
````

````python
clean_name = name.strip()
print("[" + clean_name + "]")
````

문자열 가운데의 공백은 제거하지 않습니다.

````python
message = "Hello Python"
print(message.strip())
````

### 3-2. `replace()` — 일부 문자열 바꾸기

`replace(기존, 새값)`은 문자열 안의 지정한 부분을 다른 문자열로 바꾼 결과를 반환합니다.

````python
phone = "010-1234-5678"
clean_phone = phone.replace("-", "")
print(clean_phone)
````

````python
sentence = "I like Java"
new_sentence = sentence.replace("Java", "Python")
print(new_sentence)
````

`replace()`도 원래 문자열 자체를 바꾸지 않습니다.

````python
print(sentence)
````

### 3-3. `lower()`와 `upper()` — 대소문자 통일

영문 데이터의 대소문자가 제각각이면 비교하기 어려울 수 있습니다. `lower()`는 모두 소문자로, `upper()`는 모두 대문자로 바꾼 새 문자열을 반환합니다.

````python
email = "Student@KHU.AC.KR"
print(email.lower())
````

````python
code_text = "ab12"
print(code_text.upper())
````

### 3-4. `split()` — 문자열을 여러 조각으로 나누기

`split(구분자)`는 하나의 문자열을 구분자를 기준으로 나눕니다. 반환 결과는 문자열이 아니라 **리스트(list)**입니다.

````python
data = "노트북,1200000,2"
items = data.split(",")
print(items)
````

````python
print(type(items))
````

````python
print(items[0])
print(items[1])
print(items[2])
````

구분자를 지정하지 않으면 연속된 공백을 기준으로 나눕니다.

````python
scores_text = "85 90 78"
scores_parts = scores_text.split()
print(scores_parts)
````

### 3-5. `in`과 `not in` — 포함 여부 확인

`in`은 특정 문자열이 포함되어 있는지 확인합니다. 결과는 `True` 또는 `False`입니다.

````python
email = "student@khu.ac.kr"
print("@" in email)
print("gmail" in email)
````

````python
print("naver" not in email)
````

Week 2에서 배운 비교 연산과 마찬가지로 결과가 `bool`이므로, 다음 주 조건문에서 판단 기준으로 사용할 수 있습니다.

## 4. 리스트 (List)

리스트는 여러 개의 값을 하나의 변수에 순서대로 저장하는 자료형입니다. 관련된 값을 하나의 묶음으로 다룰 때 사용합니다.

### 4-1. 리스트 생성

대괄호 `[]` 안에 값을 쉼표로 구분하여 작성합니다.

````python
scores = [85, 90, 78]
print(scores)
print(type(scores))
````

````python
names = ["Kim", "Lee", "Park"]
print(names)
````

````python
empty = []
print(empty)
print(len(empty))
````

서로 다른 자료형을 한 리스트에 넣는 것도 문법상 가능하지만, 보통은 같은 의미를 가진 값들을 묶는 것이 코드를 이해하기 쉽습니다.

````python
mixed = [1, "Python", True]
print(mixed)
````

### 4-2. 리스트의 길이 `len()`

`len()`은 리스트의 원소 개수를 반환합니다.

````python
scores = [85, 90, 78]
print(len(scores))
````

### 4-3. 리스트 인덱싱

문자열과 마찬가지로 리스트도 0부터 시작하는 인덱스를 사용합니다.

````python
scores = [85, 90, 78, 92]
print(scores[0])
print(scores[1])
````

````python
print(scores[-1])
````

### 4-4. 리스트 슬라이싱

리스트에서도 `[시작:끝]` 문법을 그대로 사용할 수 있습니다. 결과는 새로운 리스트입니다.

````python
print(scores[1:3])
````

````python
print(scores[:2])
print(scores[-2:])
````

### 4-5. 리스트는 직접 수정할 수 있다

문자열과 달리 리스트는 각 위치의 값을 바꿀 수 있는 **가변(mutable)** 자료형입니다.

````python
scores = [85, 90, 78]
scores[0] = 100
print(scores)
````

### 4-6. `append()` — 맨 뒤에 값 추가

`append(값)`은 기존 리스트의 맨 뒤에 값을 추가합니다.

````python
scores = [85, 90, 78]
scores.append(92)
print(scores)
````

문자열 메서드 대부분은 새 문자열을 반환했지만, `append()`는 기존 리스트 자체를 변경합니다.

### 4-7. `insert()` — 원하는 위치에 값 추가

`insert(인덱스, 값)`은 지정한 위치에 값을 삽입합니다. 기존 원소들은 뒤로 밀립니다.

````python
scores = [85, 90, 78]
scores.insert(1, 100)
print(scores)
````

### 4-8. `remove()` — 값을 찾아 삭제

`remove(값)`은 리스트에서 해당 값을 찾아 삭제합니다.

````python
items = ["A", "B", "C", "D"]
items.remove("B")
print(items)
````

같은 값이 여러 번 있으면 첫 번째로 발견한 값 하나를 삭제합니다.

````python
items = ["A", "B", "B", "C"]
items.remove("B")
print(items)
````

### 4-9. `pop()` — 값을 꺼내면서 삭제

`pop()`은 마지막 원소를 삭제하면서 그 값을 반환합니다. `pop(인덱스)`를 사용하면 특정 위치의 값을 꺼낼 수 있습니다.

````python
items = ["A", "B", "C"]
last = items.pop()
print(last)
print(items)
````

````python
items = ["A", "B", "C"]
first = items.pop(0)
print(first)
print(items)
````

### 4-10. 숫자 리스트 요약하기

숫자 리스트에는 다음 기본 함수를 사용할 수 있습니다.

- `sum(리스트)`: 합계
- `min(리스트)`: 최솟값
- `max(리스트)`: 최댓값
- `len(리스트)`: 원소 개수

````python
scores = [85, 90, 78, 92]
print(sum(scores))
print(min(scores))
print(max(scores))
print(len(scores))
````

````python
average = sum(scores) / len(scores)
print(average)
````

### 4-11. 리스트에서 `in`과 `not in`

리스트에서는 값이 원소로 존재하는지 확인합니다.

````python
names = ["Kim", "Lee", "Park"]
print("Lee" in names)
print("Choi" in names)
````

````python
print("Choi" not in names)
````

## 5. 입력 데이터 구조화

Week 2에서 `input()`은 항상 문자열을 반환한다고 배웠습니다. 이번에는 사용자가 여러 값을 한 번에 입력했을 때, `split()`과 리스트를 이용해 각 값을 분리합니다.

### 5-1. `input()`으로 여러 값 받기

````python
# 직접 실행할 때 85 90 78을 입력해 보세요.
parts = input("점수 세 개를 입력하세요: ").split()
print(parts)
````

### 5-2. `split()` 결과의 원소는 문자열

`split()`으로 나눈 뒤에도 각 원소의 자료형은 `str`입니다.

````python
print(parts[0])
print(type(parts[0]))
````

따라서 숫자 계산에 사용하려면 각 원소를 숫자로 변환해야 합니다.

````python
scores = [int(parts[0]), int(parts[1]), int(parts[2])]
print(scores)
print(type(scores[0]))
````

````python
print(sum(scores) / len(scores))
````

### 5-3. 쉼표로 구분된 데이터 처리

실제 데이터는 다음처럼 쉼표나 다른 기호로 구분되어 있을 수 있습니다.

````python
record = "노트북,1200000,2"
parts = record.split(",")
print(parts)
````

````python
product = parts[0]
price = int(parts[1])
quantity = int(parts[2])

print(product)
print(price)
print(quantity)
````

````python
total = price * quantity
print(f"총액: {total}원")
````

### 예제 4. 상품 주문 데이터 처리

사용자가 `상품명,가격,수량` 형식으로 입력한다고 가정합니다.

입력 예:
```text
노트북,1200000,2
```

아래 조건을 만족하도록 코드를 작성하세요.

1. `input()`으로 한 줄을 입력받습니다.
2. 쉼표를 기준으로 분리합니다.
3. 가격과 수량은 정수로 변환합니다.
4. 총액을 계산합니다.
5. 다음 형식으로 출력합니다.


### 예제 5. 세 과목 점수와 평균

공백으로 구분한 세 과목 점수를 한 번에 입력받아 숫자 리스트로 만들고 평균을 출력하세요.

입력 예:
```text
85 90 78
```

출력 예:
```text
점수: [85, 90, 78]
평균: 84.33
```

아직 반복문을 배우지 않았으므로 세 원소를 각각 `int()`로 변환하세요.

````python
scores=(input("세 과목의 점수를 입력해주세요:"))
parts=scores.split(" ")
score1=int(parts[0])
score2=int(parts[1])
score3=int(parts[2])
scores=[score1,scores,score3]
average=(score1+score2+score3)/3
print("점수: ",scores)
print("평균: ",average)
````
