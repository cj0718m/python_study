## 1. 변수 (Variable)

변수는 값을 가리키는 이름입니다. `age = 20`이라고 쓰면 이후에는 `age`라는 이름으로 20을 사용할 수 있습니다.

Python이 다루는 숫자나 문자열 등의 값을 **객체(object)**라고 합니다. 변수에 값을 할당한다는 것은 변수 이름이 그 객체를 가리키도록 한다는 뜻입니다.

### 1-1. 변수 생성과 할당

`=`는 오른쪽 값을 왼쪽 변수에 **할당**합니다. 오른쪽에 계산식이 있으면 계산을 먼저 합니다.

`print()`는 괄호 안의 값을 화면에 출력하는 함수입니다. 함수는 이름 뒤에 괄호를 붙여 실행하며, 괄호 안에는 처리할 값을 적습니다.

```python
x = 110000
print(x)
```

위 코드에서 `10`은 직접 적은 값이고, `x`는 그 값을 다루기 위해 붙인 변수 이름입니다.
`print(x)`를 실행하면 Python은 `x`가 가리키는 값을 찾아 출력합니다.

오른쪽에는 값 하나뿐 아니라 계산식도 올 수 있습니다. 이때도 오른쪽을 먼저 계산합니다.

```python
x = 5 + 3
print(x)
```

`3 + 2`를 먼저 계산한 뒤, 결과 5를 `x`에 할당합니다.

### 1-2. 변수의 값 사용하기

`b = a`는 `a`의 현재 값을 `b`에 할당합니다. `a`의 값은 바뀌지 않습니다.

```python
a = 3
b = a
print(b)
```

오른쪽에 계산식이 있으면 변수의 현재 값으로 계산합니다.

```python
c = a + 2
print(c)
```

### 1-3. 재할당

이미 만든 변수에 새 값을 할당하면, 이후에는 새 값을 사용합니다.

```python
a = 3
a = 3.14
print(a)
```

### 1-4. 변수 이름 규칙

변수 이름에는 영문자, 숫자, 밑줄 `_`을 사용할 수 있지만 숫자로 시작할 수는 없습니다. 한글 이름도 문법상 가능합니다.

대소문자는 구분합니다. `age`와 `Age`는 서로 다른 변수입니다.

| 이름 | 사용 가능 여부 | 이유 |
|---|---|---|
| `student1`, `my_score`, `_total` | 가능 | 문자나 밑줄로 시작 |
| `123abc` | 불가능 | 숫자로 시작 |
| `my var` | 불가능 | 공백 포함 |
| `my-score` | 불가능 | `-`가 뺄셈 기호로 해석됨 |
| `a@b` | 불가능 | `@`는 이름에 사용할 수 없음 |
| `if` | 불가능 | Python의 예약어 |

`if`, `for`, `True`, `None`은 용도가 정해진 **예약어**입니다. `print`, `type`, `int`는 예약어는 아니지만 함수 이름이므로 변수 이름으로 쓰지 않는 편이 좋습니다.

```python
age = 20
Age = 30
print(age)
```

```python
print(Age)
```

```python
my_score = 95
student1 = "Moon"
print(my_score,student1)
```

```python
print(student1)
```

### 1-5. 리터럴과 표현식

**리터럴(literal)**은 코드에 직접 적은 값입니다. `42`, `1.5`, `"hello"`, `True`가 그 예입니다.

**표현식(expression)**은 실행하면 값이 되는 코드입니다. `3 + 2`, `a + 5`뿐 아니라 값 하나인 `42`나 변수 이름 `a`도 표현식입니다.

`x = 42`처럼 작업을 지시하는 코드 단위를 **문장(statement)**이라고 합니다. 이 문장에서 `42`는 표현식이고, `x = 42` 전체는 할당문입니다.

```python
x = 42
name = "Hong"
print(x)
```

```python
print(name)
```

위 코드에서 `42`와 `"Hong"`은 리터럴이고, `x`와 `name`은 변수 이름입니다.

```python
a = 3
b = a + 2
print(b)
```

`a + 2`는 표현식입니다. 먼저 계산되어 `5`가 되고, 그 결과가 `b`에 할당됩니다.

숫자의 곱셈은 `*`로 씁니다. `3 * 2`는 3과 2를 곱하는 표현식입니다.

## 2. 자료형 (Data Types)

자료형은 값의 종류입니다. 자료형에 따라 사용할 수 있는 연산과 그 결과가 달라집니다.

| 자료형 | 값의 예 | 의미 |
|---|---|---|
| `int` | `10`, `-3`, `0` | 정수 |
| `float` | `3.14`, `1.5e3` | 실수 |
| `str` | `"Python"`, `"10"` | 문자열 |
| `bool` | `True`, `False` | 참과 거짓 |
| `NoneType` | `None` | 값이 없음을 나타내는 특별한 값의 자료형 |

`None`은 자료형 이름이 아니라 값입니다.

### 2-1. `type()`으로 자료형 확인하기

`type(값)`은 값의 자료형을 알려 줍니다. `print(type(10))`에서는 안쪽의 `type(10)`으로 자료형을 구한 뒤, 바깥의 `print()`로 출력합니다.

`<class 'int'>`는 자료형이 정수형 `int`라는 뜻입니다.

```python
print(type(10))
print(type(3.14))
```

```python
print(type("hello"))
print(type(True))
print(type(None))
```

### 2-2. 정수 `int`

`int`는 양의 정수, 음의 정수, 0을 나타냅니다.

`print(a, b, c)`처럼 값을 쉼표로 구분하면 여러 값을 순서대로 출력합니다. 값 사이에는 공백이 들어갑니다.

```python
a = 10
b = -3
c = 0
```

```python
print(a, b, c)
print(type(a))
```

### 2-3. 실수 `float`

`3.14`처럼 소수점이 있는 수는 `float`입니다. `1.5e3`처럼 지수 표기법으로 적을 수도 있습니다.

- `1.5e3`: 1.5에 10의 3제곱을 곱한 값, 즉 `1500.0`
- `1.5e-2`: 1.5에 10의 -2제곱을 곱한 값, 즉 `0.015`

`e`는 별도의 변수가 아니라 숫자를 적는 방식의 일부입니다.

```python
a = 3.14
print(a)
```

```python
b = 1.5e3
print(b)
```

```python
c = 1.5e-2
print(c)
```

`1.`은 `1.0`, `.3`은 `0.3`과 같습니다. 소수점 앞뒤의 0을 생략한 표기입니다.

```python
print(type(1.0))
print(type(1.))
```

```python
print(.3)
print(type(.3))
```

### 2-4. `/`와 `//`

`/`는 나눗셈입니다. 정수끼리 나누어도 결과는 `float`입니다.

`//`는 나눈 값을 **내림**합니다. 내림은 계산한 값보다 크지 않은 가장 가까운 정수를 구하는 것입니다. `5 / 2`는 2.5이고, `5 // 2`는 2입니다.

```python
print(5 / 2)
print(type(5 / 2))
```

```python
print(5 // 2)
print(type(5 // 2))
```

```python
print(4 / 2)
print(4 // 2)
```

`4 / 2`는 나누어떨어져도 `2.0`입니다. 정수끼리 `//`로 나누면 정수 결과가 나옵니다. 다만 `5.0 // 2`처럼 실수가 포함되면 결과도 `2.0`인 실수입니다.

```python
print(5.0 // 2)
print(type(5.0 // 2))
```

음수를 내리면 더 작은 정수로 갑니다. -2.5를 내린 값은 -3이므로 `-5 // 2`는 -3입니다.

```python
print(-5 / 2)
print(-5 // 2)
```

### 소수 계산의 오차

컴퓨터는 `float`를 이진수로 저장합니다. 0.1과 같은 수는 이진수로 적으면 소수 부분이 끝나지 않아, 저장 가능한 자릿수에 맞춘 가까운 값으로 저장됩니다. 이 값으로 계산하면 결과에도 작은 오차가 남을 수 있습니다.

### 2-5. 문자열 `str`

문자열은 텍스트를 나타내는 자료형입니다. 문자열 리터럴은 홑따옴표 `'` 또는 겹따옴표 `"`로 감쌉니다.

```python
s1 = 'Python'
s2 = "파이썬"
print(s1)
```

```python
print(s2)
```

따옴표 모양이 다르더라도 두 값은 모두 `str`입니다.

```python
print(type('Python'))
print(type("Python"))
```

문자열에서 `+`는 연결, 문자열과 정수 사이의 `*`는 반복을 뜻합니다.

```python
print("Hello" + " " + "Python")
```

```python
print("ha" * 3)
```

`3 + 2`는 숫자의 덧셈이고, `"3" + "2"`는 문자열의 연결입니다.

```python
print(3 + 2)
print("3" + "2")
```

### 2-6. 이스케이프 문자

문자열 안에서 백슬래시 `\`와 특정 문자를 함께 쓰면 줄바꿈 등을 나타낼 수 있습니다.

- `\n`: 줄바꿈
- `\t`: 탭 간격

문자 `n`이나 `t`를 그대로 출력하는 것과 다릅니다.

```python
print("안녕\n세상")
```

```python
print("이름:\t홍길동")
```

문자열을 감싸는 따옴표와 같은 따옴표를 내용에 넣으려면 앞에 `\`를 붙입니다. `\"`는 문자열의 끝이 아니라 내용에 포함된 큰따옴표라는 뜻입니다.

```python
print("He said, \"Hello.\"\n'It\'s Python.'")
```

백슬래시 자체를 문자로 출력하려면 `\\`처럼 두 번 적습니다.

### 2-7. 여러 줄 문자열

따옴표를 세 개 연속으로 사용하면 줄바꿈을 포함한 문자열을 작성할 수 있습니다.

```python
text = """첫 번째 줄
두 번째 줄
세 번째 줄"""
```

### 2-8. 논리값 `bool`

`bool`은 참과 거짓을 나타내는 자료형입니다. `True`는 참, `False`는 거짓이며 첫 글자는 대문자로 씁니다.

비교 연산의 결과도 논리값입니다. `3 > 2`는 3이 2보다 큰지, `3 < 2`는 3이 2보다 작은지 비교합니다.

```python
a = True
b = False
print(a)
```

```python
print(b)
print(type(a))
```

```python
print(3 > 2)
print(3 < 2)
```

Python에서는 `True`가 숫자 연산에서 1, `False`가 0처럼 동작할 수 있습니다.

```python
print(True + 1)
print(False + 1)
```

### 2-9. `None`

`None`은 값이 아직 정해지지 않았거나 결과가 없음을 나타내는 특별한 값입니다.

`result = None`은 변수 `result`에 `None`을 할당한 문장입니다.

```python
result = None
print(result)
print(type(result))
```

`None`과 숫자 0은 의미가 다릅니다.
예를 들어 점수에서 `0`은 실제 점수일 수 있지만 `None`은 아직 점수가 정해지지 않았다는 상태를 나타낼 수 있습니다.

```python
score_a = 0
score_b = None
print(score_a)
```

```python
print(score_b)
```

초기값은 변수의 용도에 맞게 정합니다. 합계를 구하려면 0에서 시작할 수 있고, 글자를 이어 붙이려면 빈 문자열 `""`에서 시작할 수 있습니다.

`""`는 문자 수가 0인 문자열입니다. 공백 한 칸이 들어 있는 `" "`와 다릅니다. 상태를 참과 거짓으로 나타내려면 `True`나 `False`를 사용합니다.

```python
total = 0
print(total)
```

```python
text = ""
print("[" + text + "]")
```

```python
flag = False
print(flag)
```

## 3. 형 변환 (Type Casting)

형 변환은 값을 다른 자료형으로 바꾸는 과정입니다.
대표적으로 `int()`, `float()`, `str()`, `bool()`을 사용합니다.

### 3-1. 자동 변환과 명시적 변환

일부 숫자 연산에서는 Python이 자료형을 자동으로 맞춥니다.
예를 들어 `int`와 `float`을 더하면 결과는 `float`이 됩니다.

```python
result = 3 + 2.5
print(result)
print(type(result))
```

문자열과 숫자 사이에서는 `+`가 자료형을 자동으로 맞춰 주지 않습니다.

#### 자료형이 다른 값의 덧셈

`"10" + 5`는 문자열과 정수를 더하려 하므로 `TypeError`가 발생합니다. `TypeError`는 해당 자료형 조합으로는 그 연산을 할 수 없다는 오류입니다.

숫자 덧셈을 하려면 `"10"`을 정수로 바꾸고, 문자열 연결을 하려면 `5`를 문자열로 바꿔야 합니다.

### 3-2. 문자열을 숫자로 변환하기

숫자 형태의 문자열은 `int()` 또는 `float()`으로 숫자형으로 바꿀 수 있습니다.

```python
x = int("123")
print(x)
print(type(x))
```

```python
y = float("3.14")
print(y)
print(type(y))
```

`int()`에 문자열을 넣을 때는 문자열이 정수 형태여야 합니다. `"123"`은 가능하지만 `"hello"`와 `"3.14"`는 정수 형태가 아닙니다.

#### 변환할 수 없는 값

`int("hello")`와 `int("3.14")`는 모두 `ValueError`가 발생합니다. `ValueError`는 자료형 자체는 받을 수 있지만, 들어 있는 값이 해당 작업에 맞지 않는다는 뜻입니다.

`"3.14"`를 실수로 읽으려면 `float("3.14")`를 사용합니다.

### 3-3. 숫자를 문자열로 변환하기

`str()`은 숫자 등을 문자열로 바꿉니다.

```python
a = str(42)
b = str(3.14)
print(a, type(a))
```

```python
print(b, type(b))
```

### 3-4. 숫자를 `bool`로 변환하기

숫자 0은 `False`, 0이 아닌 숫자는 `True`로 변환됩니다.

```python
print(bool(0))
print(bool(1))
```

```python
print(bool(-3))
print(bool(2.5))
```

### 실수를 정수로 변환하기

`int()`에 실수를 넣으면 소수 부분을 버립니다. 반올림하지 않습니다.

```python
print(int(12.7))
print(int(3.2))
```

음수에서도 0 방향으로 소수 부분을 제거합니다.

```python
print(int(-12.7))
```

`int(-12.7)`은 -12입니다. 소수 부분을 버리는 `int()`와 더 작은 정수로 내리는 `//`는 음수에서 결과가 다를 수 있습니다.

```python
print(int(-5 / 2))
print(-5 // 2)
```

## 4. 연산자 (Operators)

연산자는 계산이나 비교를 수행하는 기호 또는 단어입니다.

### 4-1. 산술 연산자

| 연산자 | 의미 | 예 | 결과 |
|---|---|---|---|
| `+` | 덧셈 | `3 + 2` | `5` |
| `-` | 뺄셈 | `3 - 2` | `1` |
| `*` | 곱셈 | `3 * 2` | `6` |
| `/` | 나눗셈 | `5 / 2` | `2.5` |
| `//` | 나눗셈 결과의 내림 | `5 // 2` | `2` |
| `%` | 나머지 | `5 % 2` | `1` |
| `**` | 거듭제곱 | `2 ** 10` | `1024` |

```python
print(3 + 2)
print(3 - 2)
print(3 * 2)
```

```python
print(5 / 2)
print(5 // 2)
print(5 % 2)
```

```python
print(2 ** 10)
```

`%`는 나눗셈의 나머지를 구합니다. 정수를 2로 나눈 나머지는 짝수이면 0, 홀수이면 1입니다.

```python
print(8 % 2)
print(9 % 2)
```

### 4-2. 할당 연산자

`a = a + 3`에서는 오른쪽의 기존 `a` 값에 3을 더한 다음, 그 결과를 다시 `a`에 할당합니다. `a`가 10이었다면 새 값은 13입니다.

숫자 계산에서 `a += 3`은 `a = a + 3`을 줄여 쓴 형태입니다.

```python
a = 10
a = a + 3
print(a)
```

```python
a = 10
a += 3
print(a)
```

다른 산술 연산에도 축약형이 있습니다.

| 축약형 | 풀어 쓴 형태 |
|---|---|
| `a -= 3` | `a = a - 3` |
| `a *= 3` | `a = a * 3` |

### 4-5. 논리 연산자

논리값을 `and`, `or`, `not`으로 조합할 수 있습니다.

- `not`: 참과 거짓을 반대로 바꿈

| A | B | `A and B` | `A or B` |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

```python
a = True
b = False
```

```python
print(a and b)
```

```python
print(a or b)
```

```python
print(not a)
```

```python
print(1 < x and x < 10)
```

먼저 `1 < x`와 `x < 10`을 각각 비교합니다. `x`가 7이면 둘 다 `True`이므로 최종 결과도 `True`입니다.

```python
age = 22
has_ticket = True
print(age >= 19 and has_ticket)
```

`age >= 19`와 `has_ticket`이 모두 참인지 판단합니다. `has_ticket`에는 이미 논리값이 저장되어 있으므로 그 자체를 조건으로 사용할 수 있습니다.

### 5-2. `sep=`

`print()`에 여러 값을 전달하면 값 사이에 공백이 들어갑니다. `sep`는 그 사이에 넣을 구분자입니다.

`sep="-"`는 `print()`의 괄호 안에서 구분자를 지정하는 **키워드 인자**입니다. 함수를 실행할 때 설정의 이름과 값을 함께 전달하는 방식입니다.

```python
print("A", "B", "C")
```

```python
print("A", "B", "C", sep="-")
```

`sep=""`는 구분자를 빈 문자열로 지정하므로 값 사이에 아무 문자도 넣지 않습니다.

### 5-3. `end=`

`end`는 한 번의 출력이 끝난 뒤 붙일 문자입니다. 기본값은 줄바꿈 문자 `"\n"`입니다.

`sep`가 여러 값 **사이**에 들어가는 문자라면, `end`는 출력 **끝**에 붙는 문자입니다.

```python
print("Hello")
print("World")
```

```python
print("Hello", end=" ")
print("World")
```

`end=" "`를 지정한 첫 번째 출력은 줄바꿈 대신 공백으로 끝납니다. 따라서 같은 셀의 두 번째 출력이 그 뒤에 이어집니다.

### 5-5. f-string

f-string은 변수 값을 문자열 안에 넣어 출력할 때 사용하는 방식입니다.
문자열 앞에 `f`를 붙이고, 삽입할 변수나 표현식을 `{}` 안에 적습니다.

```python
name = "홍길동"
age = 21
```

```python
print("이름: {name}, 나이: {age}")
```

문자열 앞에 `f`가 없으면 `{name}`과 `{age}`가 글자 그대로 출력됩니다. `f`를 붙이면 중괄호 안의 변수 값이 들어갑니다.

```python
print(f"이름: {name}, 나이: {age}")
```

중괄호 안에는 변수뿐 아니라 계산식도 넣을 수 있습니다. Python이 중괄호 안을 먼저 계산한 뒤 문자열에 넣습니다.

```python
age = 21
print(f"내년 나이: {age + 1}")
```

`{score:.1f}`처럼 출력 형식을 지정할 수도 있습니다.

- `score`: 출력할 값
- `:`: 출력 형식 지정의 시작
- `.1f`: 소수점 아래 한 자리까지 표시

표시할 자릿수에 맞춰 반올림하지만, 변수에 저장된 값 자체는 바꾸지 않습니다.

```python
score = 87.56
print(f"점수: {score:.1f}점")
```

```python
print(score)
```

출력은 87.6이지만 `score`에 저장된 값은 여전히 87.56입니다. `.2f`로 쓰면 소수점 아래 두 자리까지 표시합니다.
